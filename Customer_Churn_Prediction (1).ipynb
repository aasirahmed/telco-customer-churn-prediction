```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
import joblib
```

```python
url = 'https://raw.githubusercontent.com/IBM/telco-customer-churn-on-icp4d/master/data/Telco-Customer-Churn.csv'
df = pd.read_csv(url)
df.head()
```

```python
# df.shape
# df.info()
# df.columns
df["Churn"].value_counts()
```

```python
x = df.drop(['Churn', 'customerID'], axis=1)
y = df['Churn']

x.head()
# y.head()
# x.dtypes
```

Converting dtype of TotalCharges from object to numeric

```python
x['TotalCharges'] = pd.to_numeric(x['TotalCharges'], errors = 'coerce')
# x["TotalCharges"].isnull().sum()
# x[x["TotalCharges"].isnull()]
# x["TotalCharges"].dtype
x[x["TotalCharges"].isnull()][["tenure", "MonthlyCharges", "TotalCharges"]]
```

To fill values for Null columns from NaN --> 0

```python
x['TotalCharges'] = x['TotalCharges'].fillna(0)
```

Convert Y (Churn) from yes/no --> 0/1

```python
y = y.map({'Yes' : 1, 'No' : 0})
y.head()
```

one-hot Encoding

```python
x = pd.get_dummies(x, drop_first=True)
x.head()
```

Splitting data for train and test

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.2, random_state=42)
y_train.shape
```

Scaling the data

```python
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

Logistic regression

```python
model = LogisticRegression()
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
y_pred[:5]
```

Accuracy of the prediction

```python
accuracy = accuracy_score(y_test, y_pred)
# print("Accuracy :",accuracy)


# print(y_test.shape)
# print(y_pred.shape)
# print((y_test.values == y_pred).sum())

# print(y_test.value_counts())
# print(pd.Series(y_pred).value_counts())
# print(model.score(x_test, y_test))
# "Churn" in x.columns
# print(x.shape)
# print(y.shape)
# print(y.value_counts())
# print(model.coef_)
# print(confusion_matrix(y_test, y_pred))

y_prob = model.predict_proba(x_test)
y_prob[:5]
```
